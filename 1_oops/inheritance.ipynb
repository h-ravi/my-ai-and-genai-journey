```
class Person:
    def hello(self):
        print("Hello! Developer")

class Customer(Person):
    def __init__(self,name):
        self.name = name
        
    def greet(self):
        print(f"Hello! {self.name}")

class VIPCustomer(Customer):    
    def vip(self):
        print(f"VIP Hello! {self.name}")

hel = VIPCustomer("Dev")
hel.hello() # VIPCustomer ->x->Customer->x->Person->ok->print
hel.greet() # VIPCustomer ->x->Customer->ok->print
hel.vip() # VIPCustomer ->ok->print
```

```
# super()
class Clacu:
    def __init__(self, add2):
        self.add2 = add2
class aadi(Clacu):
    def __init__(self,add1,add2):
        super().__init__(add2)
        self.add1 = add1
    def add(self):
        print(f"addition of {self.add1} and {self.add2} = {self.add1+self.add2}")

add = aadi(3,4)
add.add()
```

```
# super()
class Clacu:
    def __init__(self, add2):
        self.add2 = add2
class aadi(Clacu):
    def __init__(self,add1,add2):
        super().__init__(add2)
        self.add1 = add1
    def add(self,add3):
        self.add3 = add3
        print(f"addition of {self.add1}, {self.add2} and {self.add3} = {self.add1+self.add2+self.add3}")

add = aadi(3,4)
add.add(3)
```

```
# types of inheritance
# 1. single Inheritance : 1 parent 1 child
class Person:
    def name(self):
        print("Your Name Is Dev")
class info(Person):
    def age(self):
        print("Your Age is : 20")
my = info()
my.name() # parent
my.age() # child
```

```
# 2. Multiple Inheritance : 1 chile inherit multiple Parent
class Father:
    def skills(self):
        print("Farming")
class Mother:
    def talent(self):
        print("Cooking")
class Child(Father, Mother):
    pass

my = Child()
my.skills()
my.talent()
```

```
# 3. multilevel Inheritance : parent -> child -> granchild
class A:
    def a(self):
        print("I am A")
class B(A):
    def b(self):
        print("I am B")
class C(B):
    def c(self):
        print("I am C")
d = C()
d.a()
d.b()
d.c()
```

```
# 4. Hierarchical Inheritance : multiple child inheri one parent
class I:
    def a(self):
        print("I am A")
    def b(self):
        print("I am B")
    def c(self):
        print("I am C")
class A(I):
    pass
class B(A):
    pass
class C(A):
    pass
a = A()
a.a()
b = B()
b.b()
c = C()
c.c()
```
