```
class Point:
    def __init__(self,x):
        self.x = x
p1 = Point(20)
p2 = Point(30)
print(p1+p2)
```

```
class Point:
    def __init__(self,x):
        self.x = x
    def __add__(self,other): # overloading
        return Point(self.x+other.x)
p1 = Point(20)
p2 = Point(30)
p3 = p1+p2
print(p3.x)
```

```
# __str__()
class Point:
    def __init__(self,x):
        self.x = x
    def __add__(self,other): # overloading
        return Point(self.x+other.x)
p1 = Point(20)
p2 = Point(30)
print(p1+p2)
```

```
# __str__()
class Point:
    def __init__(self,x):
        self.x = x
    def __add__(self,other): # overloading
        return Point(self.x+other.x)
    def __str__(self): # __str__()
        return f"{self.x}"
p1 = Point(20)
p2 = Point(30)
print(p1+p2)
```
