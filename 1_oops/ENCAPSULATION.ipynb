```
class BankAccount:
    def __init__(self, balance):
        self.balance = balance
acc1 = BankAccount(500)
print(acc1.balance) # anyone change
acc1.balance = 600
print(acc1.balance)
```

```
class BankAccount:
    def __init__(self, balance):
        self._balance = balance
acc1 = BankAccount(500) # anyone change
print(acc1._balance)
acc1._balance = 600
print(acc1._balance)
```

```
# the real encapsulation
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance
acc1 = BankAccount(500)
print(acc1.__balance) # not access directly through error
```

```
# the real encapsulation always use double underscore
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance
    def check_balance(self):
        print(self.__balance)
    def withdraw(self,amount):
        self.__balance -= amount
        print(f"withdraw Success. current balance is")
acc1 = BankAccount(500)
acc1.check_balance()
acc1.__balance = 600 # not change
acc1.check_balance() # print first given value 500
acc1.withdraw(100)
print("After withdraw")
acc1.check_balance()
```

```
# encapsulation not provide 100% security
# in encapsulation __balance variable name change into _BankAccount__balance (BankAccount is class name and use this variable name change data esly
# Encapsulaion is provide accidental direct access and changes
acc1.check_balance()
acc1._BankAccount__balance = 5000 # change balance
acc1.check_balance()
```
