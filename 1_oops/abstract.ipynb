```
from abc import ABC, abstractmethod
```

```
class BankApp(ABC):
    @abstractmethod
    def pay(self):
        pass
class Account(BankApp):
    def __init__(self):
        print("Account Open Successfully")
my_acc = Account()
```

```
class BankApp(ABC):
    @abstractmethod
    def pay(self): # strict rule make this method in child class
        pass
class Account(BankApp):
    def __init__(self):
        print("Account Open Successfully")
    def pay(self):
        print("Payment Successful for the account opening")
my_acc = Account()
my_acc.pay()
```
