```
# 1 method default value 0 or any
class Calculator:
    def add(self,a=0,b=0,c=0):
        return a+b+c
clac = Calculator()
print(clac.add(2,3))
print(clac.add(2,3,4))
```

```
# *arg method
class Calculator:
    def add(self,*args):
        return sum(args)
clac = Calculator()
print(clac.add(2,3))
print(clac.add(2,3,4))
```
