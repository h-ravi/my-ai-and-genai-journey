```
class instance_var:
    name = "Amit" # class variable
    def __init__(self, age):
        self.age = age # instance variable
        print(f'Hello! {self.name} your age is {age}')
n1 = instance_var(20)
n2 = instance_var(30)
```

```
# access class in second class
class Security:
    def verify(self):
        print("Password Verified")
class Bank:
    def login(self):
        security = Security() # object
        security.verify() # method call
        print("You are Login Successfully")
bank = Bank()
bank.login()
```

```
# access class in second class
class Security:
    def verify(self):
        print("Password Verified")
class Bank:
    def __init__(self):
        self.security = Security() # object
    def login(self):
        self.security.verify() # method call
        print("You are Login Successfully")
bank = Bank()
bank.login()
```
