```
for i in range(5):
    for j in range(i + 1):
        print("*", end="")
    print()
```

```
for i in range(20, 0,-1):
    for j in range(i, 0, -1):
        print("*", end="")
    print()
```

```
x = 4
for i in range(5):
    print(" "*x, end='')
    for j in range(i+1):
        print("*", end='')
    x-=1
    print()
```

```
x = 4
for i in range(10):
    if i % 2 != 0:
        print(" "*x, end='')
        print("*"*i, end='')
    else:
        continue
    x-=1
    print()
```

```
spaces = 4

for i in range(5):

    for j in range(spaces):
        print(" ", end="")

    for j in range(i + 1):
        print("*", end="")

    spaces -= 1
    print()
```
