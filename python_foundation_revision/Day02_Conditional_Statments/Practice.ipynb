```
def hello():
    print("hello")

print("Using and")
False and hello()

print("Using or")
True or hello()

print("Program End")
```

```
# Maan lo ek AI pipeline se alag-alag format me command aa rahe hain
command = ["move", 10, 20] 

match command:
    case ["start"]:
        print("Starting Engine...")
    case ["move", x, y]:
        print(f"Moving to coordinates ({x}, {y})")
    case ["stop", reason]:
        print(f"Stopping because: {reason}")
    case _:
        print("Invalid Command")
```
