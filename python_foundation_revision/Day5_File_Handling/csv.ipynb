```
import csv
```

```
# csv Writing
with open("students.csv","w") as file:
    write = csv.writer(file)
    # one row at one time using .writerow
    write.writerow(["roll","name","age"])
    write.writerow(["001","Dev","20"])
    # multiple rows in one time using .writerows
    write.writerows([
        ["002","Kamesh","18"],
        ["003","Lamesh","21"],
        ["004","Samesh","20"],
    ])
```

```
# CSV File Reading using .reader() give every in python list
with open("students.csv",'r') as read_file:
    read = csv.reader(read_file)
    for row in read:
        print(row)
        print(type(row))
```

```
# CSV Reading using .DictReader() and this function give every row as python dictonary
with open("students.csv",'r') as read_file:
    read_as_dict = csv.DictReader(read_file)
    for row in read_as_dict:
        print(row)
        print(type(row))
```
