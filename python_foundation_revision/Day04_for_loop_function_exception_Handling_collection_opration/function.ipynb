```
# basix syntax
def greet():
    print("Hello!")

greet()   # call kiya
greet()   # dobara call kiya
```

```
# single parameter
def greet(name):
    print(f"Hello, {name}!")

greet("Rahul")
greet("Priya")
```

```
# multiple parameters
def add(a, b):
    print(a + b)

add(5, 3)
add(10, 20)
```

```
# return 
def add(a, b):
    return a + b

result = add(5, 3)
print(result)
print(result * 2)
```

```
# default parameter value
def greet(name="Guest"):
    print(f"Hello, {name}!")

greet()
greet("Rahul")
```

```
# prectice 1
def square(n):
    return n**2
inpu = int(input("Enter you number : "))
result = square(inpu)
print(f"Square of {inpu} is {result}")
```

```
# prectice 2
def is_even(m):
    return m%2 == 0
even_input = int(input("Enter your number : "))
result = is_even(even_input)
print(result)
```

```
# prectice 3
def average(a, b, c):
    return (a+b+c)/3
num1 = int(input("Enter your 1st Number : "))
num2 = int(input("Enter your 2nd Number : "))
num3 = int(input("Enter your 3rd Number : "))
result = average(num1, num2, num3)
print(result)
```
