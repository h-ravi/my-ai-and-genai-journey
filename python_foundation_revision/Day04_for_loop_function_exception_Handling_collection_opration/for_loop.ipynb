# basic syntax
```
for item in iterable:
    #code block
```

````
fruits = ['banana','apple','mango']
for f in fruits:
    print(f)
````

## range()

````
for i in range(5):
    print(i)
````

````
for i in range(10,0,-1):
    print(i)
````

````
# enumerate
for idx, val in enumerate(fruits):
    print(idx, val)
````

````
#zip
new_list = ['jay','na_jay','jay_jay']
for per, f in zip(new_list, fruits):
    print(f'{per} eat fruit {f}')
````

````
# loop in dictonary
hyperparams = {"lr": 0.001, "batch_size": 32, "epochs": 10}
for key, value in hyperparams.items():
    print(key, "=", value)
````

````
# nested for loop
matrix = [[1,2],[3,4]]
for row in matrix:
    for val in row:
        print(val)
````
