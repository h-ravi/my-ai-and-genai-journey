# list

Ordered, changeable, duplicate allowes

```
data = [10, 20, 30]

data.append(40)        # end me ek item add
print(data)             # [10, 20, 30, 40]

data.insert(1, 15)      # specific position pe daalna
print(data)             # [10, 15, 20, 30, 40]

data.extend([50, 60])   # multiple items add (jaise 2 list jodna)
print(data)             # [10, 15, 20, 30, 40, 50, 60]
```

```
# remove in list 
data.remove(15)     # value se hatao
data.pop()          # last item hatao (aur return bhi karo)
data.pop(0)         # specific index se hatao
del data[0]         # index se delete
data.clear()        # sab kuch khali kar do
```

```
# Access karana(indexing and slicing)
data = [10, 20, 30, 40, 50]

print(data[0])      # pehla item -> 10
print(data[-1])     # aakhri item -> 50
print(data[1:3])    # slice -> [20, 30]
print(data[:3])     # shuru se 3 tak -> [10, 20, 30]
print(data[2:])     # 2 se aakhir tak -> [30, 40, 50]
```

```
# method
data = [30, 10, 50, 20]

data.sort()              # chhote se bade -> [10,20,30,50]
data.sort(reverse=True)  # bade se chhote
print(len(data))         # total items -> 4
print(max(data))         # sabse bada
print(min(data))         # sabse chhota
print(sum(data))         # sabka sum
print(30 in data)        # True/False - value hai ya nahi
```

# tuple

ordered, unchangeable

```
point = (10, 20)
colors = ("red", "green", "blue")
print(type(point))
```

```
# list vs tuple
my_list = [1, 2, 3]
my_list[0] = 100        # chal jayega, list changeable hai
print(my_list)          # [100, 2, 3]

my_tuple = (1, 2, 3)
my_tuple[0] = 100        #ERROR! Tuple change nahi ho sakta
```

```
# access
point = (10, 20, 30)
print(point[0])     # 10
print(point[-1])    # 30
print(point[0:2])   # (10, 20) - slicing bhi kaam karti hai
```

# set

unordered, unique values (duplicate not allowed)

```
nums = {10, 20, 30}
print(type(nums))
```

```
# remove duplicate automatically
nums = {1, 2, 2, 3, 3, 3, 4}
print(nums)
```

```
# convert list in set

marks = [80, 90, 80, 70, 90, 60]
unique_marks = set(marks)
print(unique_marks)
print(type(marks))
print(type(unique_marks))
```

```
# add and remove
fruits = {"apple", "banana"}

fruits.add("mango")        # ek item add
print(fruits)

fruits.remove("banana")    # specific item hataya
print(fruits)

fruits.discard("kiwi")     # remove jaisa hi, par error nahi dega agar item na ho
```

```
# not indexing 
fruits = {"apple", "banana", "mango"}
print(fruits[0])
```

```
# searching
fruits = {"apple", "banana", "mango"}
print("apple" in fruits)
```

```
# set operations 
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)   # Union
print(a & b)   # Intersection - common wale
print(a - b)   # Difference - sirf a me jo b me nahi
print(a ^ b)   # Symmetric diff - dono me jo common nahi
```

# dict

key-vale paires

```
student = {"name": "Rahul", "age": 20, "grade": "A"}
print(type(student))
```

```
# accessing in dictonary only using key
student = {"name": "Rahul", "age": 20, "grade": "A"}

print(student["name"])
print(student["age"])
```

```
print(student.get("name"))
print(student.get("city"))       # None (error nahi, khali None mil jayega)
print(student.get("city", "N/A")) # N/A (default value de sakate hai)
```

```
#add update and remove 
student = {"name": "Rahul", "age": 20}

student["grade"] = "A" #add
print(student)

student["age"] = 21  # update
print(student)

del student["grade"]  # remove
print(student)

student.pop("age") # remove (aur value return bhi karta hai)
```

```
# fir loop in dict

student = {"name": "Rahul", "age": 20, "grade": "A"}

for key in student:              # sirf keys milengi
    print(key)

for key in student.keys():       # same as upar.
    print(key)

for value in student.values():   # sirf values
    print(value)

for key, value in student.items():   # dono sath - sabse zyada use hota hai
    print(key, "->", value)
```

```
# check ki key is availale or not 
student = {"name": "Rahul", "age": 20}
print("name" in student)    # True - key check hoti hai, value nahi
print("Rahul" in student)   # False
```
