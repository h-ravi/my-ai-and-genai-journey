```
import random
```

```
# rendom number show under 1 to 10
print(random.randint(1,10))
```

```
# choice (ek collection me se koi bhi rendom vaue chunega)
list1 = ['mango','banana','apple','papaya','potato']
print(random.choice(list1))
```

```
# random 0.0 se 1.0 tak default hota hai
print(random.random())
```

```
# suffle yah mix karake fir random nikalega
card=[1,2,3,4,5,6]
random.shuffle(card)
print(card)
```
