# Mini Project 2 – ATM Machine

## Problem Statement

Create a **Simple ATM System** using a `while` loop.

The ATM should repeatedly display the following menu until the user chooses to exit.

```text
========== ATM MENU ==========
1. Check Balance
2. Deposit Money
3. Withdraw Money
4. Change PIN
5. Exit
==============================
```

### Initial Data

* Initial Balance = ₹15,000
* Initial PIN = `1234`

---

## Requirements

### 1. Check Balance

Before displaying the balance, ask the user to enter the PIN.

* If the PIN is correct, display the current balance.
* Otherwise, display **"Incorrect PIN!"**

---

### 2. Deposit Money

* Ask for the PIN.
* If the PIN is correct:

  * Ask for the deposit amount.
  * Deposit amount must be greater than `0`.
  * Update the balance.
  * Display the updated balance.
* Otherwise, display **"Incorrect PIN!"**

---

### 3. Withdraw Money

* Ask for the PIN.
* If the PIN is correct:

  * Ask for the withdrawal amount.
  * Withdrawal amount must be greater than `0`.
  * Balance must be sufficient.
  * Update the balance.
  * Display the updated balance.
* Otherwise, display **"Incorrect PIN!"**

---

### 4. Change PIN

* Ask for the current PIN.
* If the PIN is correct:

  * Ask for a new PIN.
  * Ask to confirm the new PIN.
  * If both PINs match, update the PIN.
  * Otherwise, display **"PIN Mismatch!"**
* If the current PIN is incorrect, display **"Incorrect PIN!"**

---

### 5. Exit

Display:

```text
Thank you for using our ATM.
Visit Again!
```

Then terminate the program.
