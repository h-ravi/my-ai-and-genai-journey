# Mini Project: Simple Bank Management System

### Problem Statement

Create a **Simple Bank Management System** using a `while` loop.

The program should display the following menu repeatedly until the user chooses to exit:

1. Deposit Money
2. Withdraw Money
3. Check Balance
4. Exit

### Requirements

* Start with an initial bank balance of **₹10,000**.
* If the user selects **Deposit**, ask for the deposit amount and add it to the balance.
* If the user selects **Withdraw**, ask for the withdrawal amount.

  * If the balance is sufficient, complete the transaction and update the balance.
  * Otherwise, display **"Insufficient Balance!"**
* If the user selects **Check Balance**, display the current account balance.
* If the user enters an invalid menu option, display **"Invalid Choice! Please try again."**
* The program should continue running until the user selects **Exit**.
* When the user exits, display **"Thank you for using the Bank Management System."**

### Bonus Challenge (Optional)

After every successful deposit or withdrawal:

* Display the updated account balance.
* Count the total number of transactions performed.
* Display the total transaction count before exiting the program.


```
balance = 10000
withdraw_count = 0
deposit_count = 0
while True:
    print("1. Deposit Money")
    print("2. Withdraw Money")
    print("3. Check Balance")
    print("4. Exit")
    choose = int(input("Choose youre option under 1 to 4 :"))
    if choose > 4 or choose < 1:
        print("Invalid Choice! Please try again.")
    else:
        if choose == 1:
            deposit = int(input("Enter Deposit Amount : "))
            balance+=deposit
            print(f'You Deposit amount is {deposit} and now your total balance is {balance}')
            deposit_count += 1
        elif choose == 2:
            withdraw = int(input("Enter withdraw Amount : "))
            if balance >= withdraw:
                balance-=withdraw
                withdraw_count += 1
            else:
                print(f"Insufficient Balance! you total balance is {balance}")
            print(f'You Withdraw amount is {withdraw} and now your total balance is {balance}')
        elif choose == 3:
            print(f'your present balance is {balance}')
        else:
            print("Thank you for using the Bank Management System.")
            print("Your Succully Transition count")
            print(f"Total withdraw count = {withdraw_count}")
            print(f'Total deposit count = {deposit_count}')
            break
```
