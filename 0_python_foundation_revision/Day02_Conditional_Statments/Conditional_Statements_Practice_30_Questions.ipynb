## Q1. User se age input lo. Agar age 18 ya usse zyada hai to 'Eligible to Vote' print karo, warna 'Not Eligible'.

```
age = int(input("Enter your age :"))
if age >= 18:
    print("You are Eligible to vote")
else:
    print("You are not eligible to vote")
```

## Q2. User se ek number lo. Agar number positive hai to 'Positive', negative hai to 'Negative', warna 'Zero' print karo.

```
num = int(input("Enter your number : "))

if num == 0:
    print("You enter number Zero")
elif num > 0:
    print(f"you entered positive number : {num}")
else:
    print(f"you enter negative number : {num}")
```

## Q7. User se salary lo. Agar salary >=50000 hai to 'High Income', agar >=25000 hai to 'Medium Income', warna 'Low Income'.

```
salary = int(input("Enter your salary : "))

if salary >= 50000:
    print(f"High Income : {salary}")
elif salary >= 25000:
    print(f"Medium Income : {salary}")
else:
    print(f"Low Income : {salary}")
```

## Q8. User se temperature lo. >35: Hot, 20-35: Pleasant, warna Cold print karo.

```
temp = int(input("Enter temperature : "))

if temp > 35:
    print(f"Temperature is Hot : {temp}")
elif temp < 35 and temp > 20:
    print(f"Temprature is Pleasant : {temp}")
else:
    print(f"Temprature is Cold : {temp}")
```

## Q10. User se marks lo. 90+: A, 75+: B, 60+: C, 33+: D, warna Fail.

```
marks = int(input("Enter your marks in percentage :"))

if marks >= 90:
    print(f"Grade A")
elif marks >= 75:
    print(f"Grade B")
elif marks >= 33:
    print(f"Grade D")
else:
    print(f"Fail")
```
