## Variables in Python

#### A variable is used to store data in memory

```
# Creating variables
name = "Dev"
age = 20
height = 5.8
is_student = True

print(name)
print(age)
print(height)
print(is_student)


# Python automatically decides the data type
x = 10          # Integer
y = 10.5        # Float
z = "Python"    # String

print(type(x))
print(type(y))
print(type(z))


# Multiple variable assignment
a, b, c = 10, 20, 30

print(a)
print(b)
print(c)


# Assign same value to multiple variables
x = y = z = 100

print(x)
print(y)
print(z)


# Changing the value of a variable
marks = 80
print(marks)

marks = 95      # Value updated
print(marks)


# Variable names are case-sensitive
name = "Dev"
Name = "Python"

print(name)
print(Name)


# Variable naming rules

# Valid variable names
student_name = "Aman"
studentAge = 21
_marks = 95

print(student_name)
print(studentAge)
print(_marks)

# Invalid variable names (Uncomment to see the error)

# 2name = "Dev"      # Cannot start with a number
# my-name = "Dev"    # Hyphen (-) is not allowed
# class = "Python"   # Reserved keyword cannot be used


# Using variables with f-string
name = "Dev"
course = "Python"

print(f"My name is {name}.")
print(f"I am learning {course}.")
```

## Data Types in Python

```
# Integer (int)
# Stores whole numbers
a = 10
print(a)
print(type(a))      # Output: <class 'int'>


# Float (float)
# Stores decimal numbers
b = 10.5
print(b)
print(type(b))      # Output: <class 'float'>


# String (str)
# Stores text inside single or double quotes
name = "Dev"
print(name)
print(type(name))   # Output: <class 'str'>


# Boolean (bool)
# Stores only True or False
is_student = True
print(is_student)
print(type(is_student))   # Output: <class 'bool'>


# List (list)
# Ordered, mutable (can be changed), allows duplicate values
fruits = ["Apple", "Banana", "Mango"]
print(fruits)
print(type(fruits))   # Output: <class 'list'>


# Tuple (tuple)
# Ordered, immutable (cannot be changed), allows duplicate values
colors = ("Red", "Green", "Blue")
print(colors)
print(type(colors))   # Output: <class 'tuple'>


# Set (set)
# Unordered, mutable, does not allow duplicate values
numbers = {1, 2, 3, 3, 4}
print(numbers)        # Duplicate value (3) is removed
print(type(numbers))  # Output: <class 'set'>


# Dictionary (dict)
# Stores data in key-value pairs
student = {
    "name": "Dev",
    "age": 20,
    "course": "Python"
}
print(student)
print(type(student))  # Output: <class 'dict'>


# NoneType (None)
# Represents no value or empty value
x = None
print(x)
print(type(x))        # Output: <class 'NoneType'>
```

## Type Casting in Python

```
# Implicit Type Casting (Automatic)
a = 10          # Integer
b = 5.5         # Float

c = a + b       # Python automatically converts int to float
print(c)        # Output: 15.5
print(type(c))  # Output: <class 'float'>

# Explicit Type Casting (Manual)

# String to Integer
x = "25"            # x is a string
y = int(x)          # Convert string into integer
print(y + 5)        # Output: 30

# Integer to Float
num = 10            # Integer value
f = float(num)      # Convert integer into float
print(f)            # Output: 10.0

# Float to Integer
pi = 3.14           # Float value
i = int(pi)         # Decimal part is removed
print(i)            # Output: 3

# Integer to String
age = 20            # Integer value
text = str(age)     # Convert integer into string
print(text)         # Output: "20"
print(type(text))   # Output: <class 'str'>

# Integer to Boolean
n = 1               # Non-zero value (only 1 integer number convert boolean in True)
b = bool(n)         # Non-zero becomes True
print(b)            # Output: True

# Zero to Boolean
z = 0               # Zero value
print(bool(z))      # Output: False
```

## print() and input() in Python

```
# print()
# Used to display output on the screen
print("Welcome to Python")

# Printing multiple values
name = "Dev"
age = 20
print(name, age)

# Using separator (sep)
print("Python", "Java", "C++", sep=" - ")
# Output: Python - Java - C++

# Using end
print("Hello", end=" ")
print("World")
# Output: Hello World


# input()
# Used to take input from the user
name = input("Enter your name: ")
print("Your name is:", name)

# Input is always stored as a string
age = input("Enter your age: ")
print(type(age))      # Output: <class 'str'>

# Convert input into integer
age = int(input("Enter your age: "))
print("Age:", age)
print(type(age))      # Output: <class 'int'>


# f-string
# Used to insert variables inside a string
name = "Dev"
age = 20

print(f"My name is {name}.")
print(f"I am {age} years old.")

# f-string with expression
a = 10
b = 5
print(f"Sum = {a + b}")     # Output: Sum = 15

# f-string with float formatting
pi = 3.141592
print(f"Value of Pi = {pi:.2f}")   # Output: Value of Pi = 3.14
```
