### **Mode**

- **r** : only read file (if file is not exist then through FileNotFoundError)
- **w** : only write file (if file exist then overwrite file and not exist then create)
- **a** : only write file (if file exist then append added content in last line is file or if file is not exist then create file)
- **r+** : read+write
- **w+** : write+read
- **a+** : append+read
- **rb** :read binary file (jpeg,png and etc...)
- **wb** :Write binary file
- **rb+** : read and write binary file
- **wb+** : write and read binary file
- **ab** : append in binary file
- **ab+** : append and read binary file

## Better approtch

```
# always use with 
with open("data.txt", "r") as file:
    print(file.read())
```

```
# read file using loop
with open("data.txt", "r") as file:
    for line in file:
        print(line.strip())
```

```
line = ["Amit\n","Sumit\n"]
with open("demo2.txt", "w") as file:
    file.writelines(line)
```

```
with open("demo2.txt", "w") as file:
    print("Hello", file=file)
    print("Python", file=file)
with open("demo2.txt", "r") as read:
    print(read.read())
```
