```
# write in json file
import json
```

```
student = [
    {
        "name":"Ravi",
        "Age" :25,
        "city":'Delhi'
    }
]
with open("student.json","w+") as file:
    json.dump(student,file)
    file.seek(0)
    print(file.read())
```

```
with open("student.json") as file:
    data = json.load(file)
print(data)
```

```
import json
student = [{"name": "Ravi", "Age": 25, "city": "Delhi"}]
json_string = json.dumps(student)
print(json_string)
```

```
student = [{"name": "Ravi", "Age": 25, "city": "Delhi"}] # dictonary in list
json_string = json.dumps(student, indent=4)
print(json_string)
print(type(json_string))
```

```
student = json.loads(json_string)
print(student)
print(type(student))
```
