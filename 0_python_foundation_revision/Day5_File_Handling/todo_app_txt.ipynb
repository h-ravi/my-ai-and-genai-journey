## Create ToDo App and store data in TXT file
- Add ToDo
- Show ToDo
- Delete ToDo
- clear ToDo
- clear and modify note

```
%cd python_foundation_revision/Day5_File_Handling
```

```
login = True
# show All ToDo
def all_todo():
    with open("my_todo.txt","r") as file:
        for index, todo in enumerate(file):
            print(f"{index+1}. {todo.strip()}")
# create Empty TXT File
try:
    todo = open("my_todo.txt",'x')
    todo.close()
except FileExistsError:
    pass
# main Application
while login:
    print('==========Options============')
    print('1. Add ToDo')
    print('2. Show All ToDo')
    print('3. Delete Selected ToDo')
    print('4. Clear All ToDo')
    print('5. Exit')
    print('==========End Options=========')
    try:
        print('\n')
        choose = int(input("Choose your option : "))
        print('\n')
    except ValueError:
        print("Invalid Option! Pleas Enter Only Number not latter.")
        continue
    if choose <= 0:
        print("Invalid Option! Please Choose Right Option")
# Exit
    elif choose == 5:
        print("Thanks for using my todo app")
        print("Your All Added and Active ToDo's")
        all_todo()
        login = False
# Add new Todo
    elif choose == 1:
        new_todo = input("Enter Your Task ToDo : ")
        with open("my_todo.txt", 'a') as file:
            file.write(f"{new_todo}\n")
        print(f'your new task todo {new_todo} added sucessfully.')
# Show All Todo
    elif choose == 2:
        print("All ToDo : ")
        all_todo()
# Delete Option
    elif choose == 3:
        print("Select which ToDo to delete (only choose through number) : ")
        all_todo()
        while True:
            try:
                del_opt = int(input("Please Enter ToDo Option Number for Deleting : "))
                del_opt -= 1
            except ValueError:
                print("Please Only Enter ToDo Option in number not latter.")
                continue
            with open("my_todo.txt","r") as del_todo:
                todos = del_todo.readlines()
            if len(todos) > del_opt >= 0:
                del todos[del_opt]
                with open("my_todo.txt",'w') as update_todo_file:
                    update_todo_file.writelines(todos)
                print("Your All ToDo After Deletion : ")
                all_todo()    
                break
            else:
                print("Not choose right todo number for deleting.")
# Clear All Todo list
    elif choose == 4:
        while True:
            print("Are Your Sure Clear All ToDo's. Enter 1 for Yes or Enter 0 for no")
            try:
                confirmation = int(input("Enter : "))
            except ValueError:
                print("Please Enter Only 0 or 1")
                continue
            if confirmation == 0:
                break
            elif confirmation == 1:
                with open("my_todo.txt", "w") as file:
                    file.write("")
                print("All Todo Cleared.")
                break
            else:
                print("Please Enter Only 0 or 1")
    else:
        print("Invalid Option. Choose Right Options under geven option.")
    
```
