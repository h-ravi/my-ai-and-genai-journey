```
class Animal:
    def sound(self):
        print("Animal Sound")
class Dog(Animal):
    def sound(self):
        print("Bark")
class Cat(Animal):
    def sound(self):
        print("Meow")
        
animals = [Dog(),Cat()]
for animal in animals:
    animal.sound()
```

```
# Duck Typing
class Animal:
    def sound(self):
        print("Animal Sound")
class Dog(Animal):
    def sound(self):
        print("Bark")
class Cat(Animal):
    def sound(self):
        print("Meow")
        
def ani_sound(animal):
    animal.sound()
ani_sound(Dog())
ani_sound(Cat())
ani_sound(Animal())
```
